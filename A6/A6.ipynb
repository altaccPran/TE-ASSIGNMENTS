### Importing required libraries for data analysis and visualization

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

### Importing the iris dataset

```python
df= pd.read_csv("Iris.csv")
```

### Displaying the list of features

```python
numeric_features = ["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]
class_column = "Species"

print("Features and their types:")
for col in numeric_features:
    print(f"{col}: numeric")
print(f"{class_column}: nominal / categorical")
```

### Histogram with distribution for every feature

```python
plt.figure(figsize=(12, 8))
for i, feature in enumerate(numeric_features):
    plt.subplot(2, 2, i+1)
    sns.histplot(df[feature], bins=10, kde=True)
    plt.title(f"Histogram of {feature}")
plt.tight_layout()
plt.show()
```

- `SepalLengthCm`: The distribution is bimodal with peaks around 5 and 6.5 cm.
- `SepalWidthCm`: The distribution is roughly normal, with a peak around 3 cm.
- `PetalLengthCm`: The distribution is right-skewed, with most values between 1 and 4 cm.
- `PetalWidthCm`: The distribution is right-skewed, with the highest frequency around 1 cm.

### Box Plot for every freature with outliers

```python
plt.figure(figsize=(12, 8))
for i, feature in enumerate(numeric_features):
    plt.subplot(2, 2, i+1)
    sns.boxplot(y=df[feature], showfliers=True)
    plt.title(f"Boxplot of {feature}")
plt.tight_layout()
plt.show()
```

- `SepalLengthCm`: mostly normal, a few high-value outliers
- `SepalWidthCm`: roughly normal, some low-value outliers
- `PetalLengthCm` & `PetalWidthCm`: clearly bimodal, almost no outliers
- **Overall**: Petal features separate species well; Sepals overlap more
- Outliers are minor and mostly in sepal measurements
