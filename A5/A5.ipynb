## Data Visualisation II

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Loading Dataset

```python
df = pd.read_csv("titanic.csv")
df.head()
```

### Basic Initial Survey

```python
df.info()
df.describe()
df.isnull().sum()
```

### Data Cleaning
**Handling Missing Values**
- Age → filled using median

- Embarked → filled using mode

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

### Survival Stats

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Survived", data=df)

plt.title("Survival Count (0 = No, 1 = Yes)")
plt.show()
```

### Box Plot of Age Distribution w.r.t Gender and Survival

```python
plt.figure(figsize=(10,6))

sns.boxplot(x="Sex", y="Age", hue="Survived", data=df)

plt.title("Box Plot of Age Distribution by Gender and Survival")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.legend(title="Survived (0 = No, 1 = Yes)")
plt.show()
```

## **Inferences from the Box Plot**

- **Female passengers had a higher survival rate**  
  - The survived (orange) box for females is more prominent, showing that more females survived compared to males.

- **Male passengers show higher deaths across all age groups**  
  - The non-survived (blue) box for males is larger, indicating that many male passengers did not survive.

- **Median age is similar for survived and non-survived males**  
  - For males, the median age line in both survived and non-survived groups lies around the late 20s, meaning survival was not strongly dependent on age for men.

- **Surviving females tend to be slightly younger**  
  - The median age of survived females is slightly lower compared to non-survived females, suggesting younger women had better chances.

- **Children had better survival chances**  
  - Lower age values appear more in the survived category, indicating that children were more likely to survive.

- **Older passengers are mostly in the non-survived group**  
  - Many higher-age outliers are present in the non-survived category, showing that elderly passengers had lower survival probability.

- **Age distribution is wider among males**  
  - Male passengers have more outliers and a wider spread, meaning males of many different age groups were aboard, but survival was still low.

---

### **Overall Conclusion**

- Gender played a major role in survival, with females having significantly higher survival chances.  
- Age also mattered, as children survived more, while older passengers had lower survival rates.
