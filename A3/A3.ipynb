### DSBDAL Assignment 3A

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('BankChurners.csv')
```

###Initial Survey

```python
display(df.head())
df.describe()
```

```python
df.info()
df.isnull().sum()
```

```python
summary_statistics = df.groupby("Gender")["Credit_Limit"].agg(
    Mean="mean",
    Median="median",
    Minimum="min",
    Maximum="max",
    Standard_Deviation="std"
)

summary_statistics
```

```python
plt.figure()
df.boxplot(column="Credit_Limit", by="Gender")

plt.title("Credit Limit Distribution by Gender")
plt.suptitle("")
plt.xlabel("Gender")
plt.ylabel("Credit Limit")

plt.show()
```

```python
plt.hist(df[df["Gender"] == "F"]["Credit_Limit"], alpha=0.6, label="Female")
plt.hist(df[df["Gender"] == "M"]["Credit_Limit"], alpha=0.6, label="Male")

plt.xlabel("Credit Limit")
plt.ylabel("Frequency")
plt.title("Credit Limit Distribution by Gender")
plt.legend()

plt.show()
```

```python
# Separate data
female = np.sort(df[df["Gender"] == "F"]["Credit_Limit"])
male = np.sort(df[df["Gender"] == "M"]["Credit_Limit"])

# ECDF values
y_f = np.arange(1, len(female)+1) / len(female)
y_m = np.arange(1, len(male)+1) / len(male)

# Plot
plt.figure()
plt.plot(female, y_f, label="Female")
plt.plot(male, y_m, label="Male")

plt.xlabel("Credit Limit")
plt.ylabel("Cumulative Probability")
plt.title("ECDF of Credit Limit by Gender")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.stripplot(
    x="Gender",
    y="Credit_Limit",
    data=df,
    jitter=0.35,      # spreads points horizontally
    dodge=True,
    size=4,           # point size
    alpha=0.5         # transparency to reduce overlap
)

plt.title("Spread Strip Plot of Credit Limit by Gender")
plt.xlabel("Gender")
plt.ylabel("Credit Limit")

plt.show()
```

# Descriptive Statistics - Iris Dataset

Objective

To write a Python program that displays basic statistical details such as:

- Mean

- Standard Deviation

- Percentiles (25%, 50%, 75%)

for the species:

- Iris-setosa

- Iris-versicolor

- Iris-virginica

using the **Iris CSV dataset**.

### Import Libraries:

```python
import pandas as pd
```

Explanation:

- `pandas` is used for data manipulation and statistical analysis.

### Load the Iris Dataset

```python
df = pd.read_csv("iris.csv")
df.head()
```

### Mean, Median, Min, Max & Standard Deviation (Species-wise)

```python
grouped_stats = df.groupby("species").agg(
    mean=("sepal_length", "mean"),
    std=("sepal_length", "std"),
    min=("sepal_length", "min"),
    max=("sepal_length", "max"),
    median=("sepal_length", "median")
)

grouped_stats
```

Explanation:

- The data is grouped by `species`.

- Descriptive statistics are calculated for **sepal length**.

### Percentiles, Mean & Standard Deviation for All Features

```python
percentile_stats = df.groupby("species").describe(percentiles=[0.25, 0.5, 0.75])
percentile_stats
```

Explanation:

- `describe()` generates:

  - Count

  - Mean

  - Standard deviation

  - Minimum & maximum

  - 25th, 50th (median), and 75th percentiles

- Statistics are calculated separately for each species and each numeric attribute.
