# Data Analytics III
## Naïve Bayes Classification on Iris Dataset

### Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

### Load the Dataset

```python
df = pd.read_csv("Iris.csv")
df.head()
```

### Data Preprocessing

```python
# Check dataset info
df.info()

# Drop unnecessary column (Id)
df.drop("Id", axis=1, inplace=True)

# Separate features and target
X = df.drop("Species", axis=1)
y = df["Species"]

# Train-Test Split (70-30)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

### Apply Naïve Bayes Algorithm

```python
# Create Gaussian Naïve Bayes model
model = GaussianNB()

# Train model
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)
```

### Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

### Confusion Matrix Visualisation

```python
plt.figure()
sns.heatmap(cm, annot=True, fmt="d")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Naïve Bayes")
plt.show()
```

### Accuracy vs Error Rate Visualization

```python
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy

metrics = [accuracy*100, error_rate*100]
labels = ["Accuracy (%)", "Error Rate (%)"]

plt.figure()
plt.bar(labels, metrics)
plt.ylabel("Percentage")
plt.title("Accuracy vs Error Rate")

# Add values on top
for i in range(len(metrics)):
    plt.text(i, metrics[i], f"{metrics[i]:.2f}%", ha='center')

plt.show()
```

### Precision, Recall, F1-Score Visualization (Per Class)

```python
from sklearn.metrics import classification_report

report = classification_report(y_test, y_pred, output_dict=True)
report_df = pd.DataFrame(report).transpose()

# Remove accuracy row
report_df = report_df.iloc[:-1, :]

# Select required metrics
report_df = report_df[["precision", "recall", "f1-score"]]

# Convert to percentage
report_df = report_df * 100

report_df.plot(kind="bar")
plt.ylabel("Percentage")
plt.title("Precision, Recall, F1-Score per Class")
plt.xticks(rotation=45)
plt.show()
```

### Calculate TP, FP, TN, FN
Since Iris is a **multiclass dataset (3 classes)**, TP, FP, TN, FN are calculated per class.

```python
# Extract values per class
for i in range(len(cm)):
    TP = cm[i, i]
    FN = cm[i, :].sum() - TP
    FP = cm[:, i].sum() - TP
    TN = cm.sum() - (TP + FP + FN)

    print(f"\nClass {i}:")
    print("TP =", TP)
    print("FP =", FP)
    print("FN =", FN)
    print("TN =", TN)
```

### Accuracy, Error Rate, Precision, Recall

```python
accuracy = accuracy_score(y_test, y_pred)
error_rate = 1 - accuracy

# Round to 2 decimal places
accuracy_rounded = round(accuracy, 2)
error_rate_rounded = round(error_rate, 2)

# Convert to percentage
accuracy_percent = round(accuracy * 100, 2)
error_rate_percent = round(error_rate * 100, 2)

print("Accuracy (Decimal):", accuracy_rounded)
print("Error Rate (Decimal):", error_rate_rounded)

print("Accuracy (%):", accuracy_percent, "%")
print("Error Rate (%):", error_rate_percent, "%")
```

**Detailed Metrics**

```python
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

## Expected Output Explanation

- **Accuracy** → Overall correctness of model
- **Error Rate** → 1 − Accuracy
- **Precision** → TP / (TP + FP)
- **Recall** → TP / (TP + FN)
- **Confusion Matrix** → Shows classification performance

For Iris dataset, Naïve Bayes typically gives **95%–100% accuracy**.
