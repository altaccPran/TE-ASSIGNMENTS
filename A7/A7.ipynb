## Text Analysis
### Import Required Libraries

```python
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

### Download NLTK Resources

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("averaged_perceptron_tagger")
nltk.download("punkt_tab")
nltk.download('averaged_perceptron_tagger_eng')
```

### Sample Document Input

```python
doc = """
Education plays a very important role in shaping the future of individuals and society.
It helps people gain knowledge, develop skills, and build confidence. From childhood,
students learn basic subjects like mathematics, science, language, and history, which
form the foundation for higher studies.

In today’s world, education is not limited to classrooms only. Online learning platforms,
digital libraries, and interactive courses have made knowledge accessible to everyone.
Many students also participate in extracurricular activities such as sports, music,
debates, and cultural events to improve their overall personality.

Teachers guide learners by providing support, motivation, and discipline. At the same time,
parents contribute by encouraging good habits and creating a positive learning environment
at home. Education also helps in developing moral values such as honesty, responsibility,
and respect for others.

A well-educated person can contribute to the progress of the nation by working in various
fields like medicine, engineering, business, arts, and public service. Therefore, education
is considered one of the most powerful tools for personal growth and national development.
"""

print("Sample Document:\n", doc)
```

### Document Preprocessing Methods
**Tokenization**

```python
tokens = word_tokenize(doc)
print("Tokens:\n", tokens)
```

### POS Tagging

```python
pos_tags = nltk.pos_tag(tokens)
print("POS Tags:\n", pos_tags)
```

### Stop Words Removal

```python
stop_words = set(stopwords.words("english"))
filtered_tokens = [word for word in tokens if word.lower() not in stop_words and word not in string.punctuation]
print("After Stop Word Removal:\n", filtered_tokens)
```

### Stemming

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
for w, s in zip(filtered_tokens, stemmed_words):
    print(w, ":", s)
```

### Lemmatization

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]
for w, s in zip(filtered_tokens, lemmatized_words):
    print(w, ":", s)
```

### TF and IDF Representation
**Term Frequency (TF)**

```python
from collections import Counter

tf = Counter(filtered_tokens)

print("Term Frequency:\n")
for word, freq in tf.items():
    print(word, ":", freq)
```

### TF-IDF Representation

```python
documents = [doc]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

print("TF-IDF Feature Names:\n", vectorizer.get_feature_names_out())
print("\nTF-IDF Values:\n", tfidf_matrix.toarray())
```
